## Комментарии

Для корреляционного анализу используется не только коэффициент Пирсона.
Обратите внимание на таблицу:
![corr.png](https://wiki.pmifi.ru/machine_learning/corr.png)

Если первая величина представлена бинарной переменной (или дихотомической, или альтернативной), а вторая числовая (или вещественная, или интервальная), то используется не коэффициент Пирсона, а [Бисериальный коэффициент](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.pointbiserialr.html#scipy.stats.pointbiserialr).
Все коэффициенты можно найти здесь: https://docs.scipy.org/doc/scipy/reference/stats.html
